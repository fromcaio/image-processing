## EXERCÍCIO 01
Carregar a imagem lena (images/lenaCor.bmp), convertê-la para monocromática, e aplicar a ela um filtro passa-alta obtido através da criação de uma máscara hp.

Hp = [-1 -1 -1, -1 8 -1, -1 -1 -1]

J1 = filter2(h1,I);

```python
import numpy as np
import matplotlib.pyplot as plt
import cv2

# Carregar a imagem
img = cv2.imread('../images/lenaCor.bmp')

# Converter para monocromática (escala de cinza)
if img is None:
    raise FileNotFoundError("Imagem não encontrada. Verifique o caminho.")

img_gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Definir o filtro passa-alta
hp_filter = np.array([[-1, -1, -1],
                      [-1,  8, -1],
                      [-1, -1, -1]], dtype=np.float32)

# Aplicar o filtro usando cv2.filter2D (equivalente ao filter2 do Octave/MATLAB)
# o parâmetro -1 indica que a profundidade (numero de bits de cada pixel) da imagem de saída será a mesma da imagem de entrada
# 0 poderia ser usado para converter a imagem para 8 bits mas ela já está em 8 bits
filtered_img = cv2.filter2D(img_gray, -1, hp_filter)

# Exibir as imagens
plt.figure(figsize=(12, 5))

plt.subplot(1, 3, 1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title('Imagem Original (Colorida)')
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(img_gray, cmap='gray')
plt.title('Imagem Monocromática')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(filtered_img, cmap='gray')
plt.title('Filtro Passa-Alta Aplicado')
plt.axis('off')

plt.tight_layout()
plt.show()
```

## EXERCÍCIO 02
Carregar a imagem lena, convertê-la para monocromática e aplicar a ela um filtro de Robert obtido através da criação de uma máscara Robert. 

robert=[1 0, 0 -1] 

J2 = filter2(h2,Robert);

```python
import numpy as np
import matplotlib.pyplot as plt
import cv2

# Carregar a imagem
img = cv2.imread('../images/lenaCor.bmp')

# Converter para monocromática (escala de cinza)
if img is None:
    raise FileNotFoundError("Imagem não encontrada. Verifique o caminho.")

img_gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Definir o filtro roberts
hp_filter1 = np.array([[1, 0],
                      [0, -1]], dtype=np.float32)
hp_filter2 = np.array([[0, 1],
                        [-1, 0]], dtype=np.float32)

# Aplicar o filtro usando cv2.filter2D (equivalente ao filter2 do Octave/MATLAB)
# o parâmetro -1 indica que a profundidade (numero de bits de cada pixel) da imagem de saída será a mesma da imagem de entrada
# 0 poderia ser usado para converter a imagem para 8 bits mas ela já está em 8 bits
filtered_img1 = cv2.filter2D(img_gray, -1, hp_filter1)
filtered_img2 = cv2.filter2D(img_gray, -1, hp_filter2)

filtered_img = filtered_img1 + filtered_img2

# Exibir as imagens
plt.figure(figsize=(12, 5))

plt.subplot(1, 3, 1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title('Imagem Original (Colorida)')
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(img_gray, cmap='gray')
plt.title('Imagem Monocromática')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(filtered_img, cmap='gray')
plt.title('Filtro de Roberts Aplicado')
plt.axis('off')

plt.tight_layout()
plt.show()
```

## EXERCÍCIO 03
Carregar a imagem lena, convertê-la para monocromática e aplicar a ela um filtro obtido através da criação de uma máscara de convolução por meio da função fspecial com a opção Prewitt e Sobel. 

h=fspecial('prewitt'); 

I2=filter2(h,im);

```python
import numpy as np
import matplotlib.pyplot as plt
import cv2

# Carregar a imagem
img = cv2.imread('../images/lenaCor.bmp')

# Converter para monocromática (escala de cinza)
if img is None:
    raise FileNotFoundError("Imagem não encontrada. Verifique o caminho.")

img_gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Definir os filtros Sobel
sobel_x = np.array([[-1, 0, 1],
                    [-2, 0, 2],
                    [-1, 0, 1]], dtype=np.float32)

sobel_y = np.array([[-1, -2, -1],
                    [ 0,  0,  0],
                    [ 1,  2,  1]], dtype=np.float32)

# Filtros de Prewitt (equivalente a fspecial('prewitt') do Octave)
prewitt_x = np.array([[-1, 0, 1],
                      [-1, 0, 1],
                      [-1, 0, 1]], dtype=np.float32)
prewitt_y = np.array([[-1, -1, -1],
                      [ 0,  0,  0],
                      [ 1,  1,  1]], dtype=np.float32)

# Aplicar os filtros usando cv2.filter2D
filtered_sobel_x = cv2.filter2D(img_gray, -1, sobel_x)
filtered_sobel_y = cv2.filter2D(img_gray, -1, sobel_y)
filtered_sobel = filtered_sobel_x + filtered_sobel_y

filtered_prewitt_x = cv2.filter2D(img_gray, -1, prewitt_x)
filtered_prewitt_y = cv2.filter2D(img_gray, -1, prewitt_y)
filtered_prewitt = filtered_prewitt_x + filtered_prewitt_y

# Exibir as imagens
plt.figure(figsize=(15, 8))
plt.subplot(2, 3, 1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title('Imagem Original (Colorida)')
plt.axis('off')
plt.subplot(2, 3, 2)
plt.imshow(img_gray, cmap='gray')
plt.title('Imagem Monocromática')
plt.axis('off')
plt.subplot(2, 3, 3)
plt.imshow(filtered_sobel, cmap='gray')
plt.title('Filtro de Sobel Aplicado')
plt.axis('off')
plt.subplot(2, 3, 5)
plt.imshow(filtered_prewitt, cmap='gray')
plt.title('Filtro de Prewitt Aplicado')
plt.axis('off')
plt.tight_layout()
plt.show()
```

## EXERCÍCIO 04
Carregar a imagem cameraman.tif e crie um versão levemente borrada utilizando. 

g = fspecial("gaussian",[3,3],1.5); 
imgFiltrado = imfilter(img,g); 

Em seguida utilize o filtro laplaciano e crie uma nova imagem subtraindo a versão levemente borrada da imagem filtrado com o filtro laplaciano. 

filtro_laplaciano = [0 1 0; 1 -4 1; 0 1 0]; 
bordas = imfilter(imgFiltrado, filtro_laplaciano); 

Visualizar as imagens na mesma janela.

```python
import numpy as np
import matplotlib.pyplot as plt
import cv2

# Carregar a imagem cameraman.tif
img = cv2.imread('../images/cameraman.tif', cv2.IMREAD_GRAYSCALE)

# Verificar se a imagem foi carregada
if img is None:
    raise FileNotFoundError("Imagem cameraman.tif não encontrada. Verifique o caminho.")

# Converter para float32 para evitar problemas de overflow
img_float = img.astype(np.float32)

# 1. Criar versão levemente borrada usando filtro Gaussiano
# Equivalente ao fspecial("gaussian",[3,3],1.5) do Octave
# 3 - tamanho do kernel
# 1.5 - desvio padrão (sigma)
gaussian_kernel = cv2.getGaussianKernel(3, 1.5)
# gaussian_kernel.T é a transposta do kernel 1D
# Aqui estamos transformando um filtro 1D (ou seja só possui uma coluna) em um filtro 2D
# Desse modo borramos tanto na horizontal quanto na vertical
gaussian_filter = gaussian_kernel * gaussian_kernel.T
# Aplicar o filtro Gaussiano
img_blurred = cv2.filter2D(img_float, -1, gaussian_filter)

# 2. Definir o filtro Laplaciano
# filtro_laplaciano = [0 1 0; 1 -4 1; 0 1 0]
laplacian_filter = np.array([[0,  1,  0],
                            [1, -4,  1],
                            [0,  1,  0]], dtype=np.float32)

# 3. Aplicar o filtro Laplaciano na imagem borrada
bordas_laplacian = cv2.filter2D(img_blurred, -1, laplacian_filter)

# 4. Criar nova imagem subtraindo as bordas da imagem borrada
# (Esta é a técnica de "unsharp masking")
img_sharpened = img_blurred - bordas_laplacian

# Garantir que os valores estejam no intervalo [0, 255]
img_sharpened = np.clip(img_sharpened, 0, 255).astype(np.uint8)
img_blurred = np.clip(img_blurred, 0, 255).astype(np.uint8)
bordas_laplacian = np.clip(np.abs(bordas_laplacian), 0, 255).astype(np.uint8)

# Visualizar as imagens na mesma janela
plt.figure(figsize=(15, 10))

plt.subplot(2, 3, 1)
plt.imshow(img, cmap='gray')
plt.title('Imagem Original (Cameraman)')
plt.axis('off')

plt.subplot(2, 3, 2)
plt.imshow(img_blurred, cmap='gray')
plt.title('Imagem Borrada (Filtro Gaussiano)')
plt.axis('off')

plt.subplot(2, 3, 3)
plt.imshow(bordas_laplacian, cmap='gray')
plt.title('Bordas (Filtro Laplaciano)')
plt.axis('off')

plt.subplot(2, 3, 4)
plt.imshow(img_sharpened, cmap='gray')
plt.title('Imagem Aguçada (Original - Laplaciano)')
plt.axis('off')

# Mostrar a diferença entre original e aguçada
difference = cv2.absdiff(img, img_sharpened)
plt.subplot(2, 3, 5)
plt.imshow(difference, cmap='gray')
plt.title('Diferença (Original vs Aguçada)')
plt.axis('off')

# Comparação lado a lado
plt.subplot(2, 3, 6)
comparison = np.hstack((img, img_sharpened))
plt.imshow(comparison, cmap='gray')
plt.title('Comparação: Original | Aguçada')
plt.axis('off')

plt.tight_layout()
plt.show()
```
