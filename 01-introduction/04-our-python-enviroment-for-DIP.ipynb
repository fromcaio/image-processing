## Our Core Libraries

We will rely on five key tools in this course:

1.  **Jupyter Notebook:** This interactive environment you're in right now! It's perfect for learning because it allows us to mix rich text explanations (like this!), executable code, and visual outputs (images and plots) all in one document.

## ⚠️ The Most Common "Gotcha": BGR vs. RGB!

This is the most common bug that new DIP students encounter. Please read this carefully.

The universal standard for digital color is **RGB** (Red, Green, Blue). This means the 1st channel is Red, the 2nd is Green, and the 3rd is Blue.

* **Matplotlib** (`plt.imshow()`) *expects* images in **RGB** format.
* **scikit-image** (`skimage.io.imread()`) loads images in **RGB** format.

However, for historical reasons (dating back to early camera manufacturers), **OpenCV (`cv2.imread()`) loads images in BGR format** (Blue, Green, Red). The 1st and 3rd channels are swapped!

**What happens if you mix them up?**

If you load an image with `cv2.imread()` (getting a **BGR** image) and display it with `plt.imshow()` (which *thinks* it's an **RGB** image), the colors will be completely wrong. It will swap all the blue and red components, making skin look blue and skies look red.

### The Solution: Always Convert!
