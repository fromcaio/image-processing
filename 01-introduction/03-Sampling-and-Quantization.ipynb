## Motivation: How to Digitize `f(x, y)`?

We now know a continuous image is a function `f(x, y)` and a digital image is a matrix `f[x, y]`. How do we *build* this matrix from the continuous function?

This conversion is called **digitization**, and it involves two independent processes:

1.  **Sampling:** Digitizing the **spatial coordinates** (`x`, `y`).
2.  **Quantization:** Digitizing the **intensity value** (`f`).

Let's look at each one.

## Theory: Sampling (Digitizing Space)

**Sampling** is the process of taking measurements of the continuous image at discrete, regular intervals. 

Imagine laying a grid over the continuous image. You then check the intensity value *only* at the center of each grid square. The result of this process gives you the *dimensions* of your image matrix, `M x N`.

### Spatial Resolution
The density of this grid is the **Spatial Resolution**. 
* A **high resolution** (e.g., 4000 x 3000) means you used a very dense grid with many samples. This captures fine details.
* A **low resolution** (e.g., 64 x 64) means you used a sparse grid with few samples. This loses detail and can result in a "blocky" or "pixelated" image.

Spatial resolution is often measured in **dpi** (dots per inch), which is common in the printing industry.

### Visual Example: The Effect of Poor Sampling

What happens when we reduce the number of samples (M and N)? The image becomes coarser and coarser until the objects are no longer recognizable.

## Theory: Quantization (Digitizing Intensity)

After sampling, we have a grid of `M x N` samples. But each sample *still* holds a continuous intensity value (e.g., 0.651... on a scale of 0 to 1).

**Quantization** is the process of mapping each continuous intensity value to a discrete integer from a fixed set of levels.

### Bit Depth (Intensity Resolution)
The number of available levels is determined by the **bit depth**, `k`. The number of levels, `L`, is given by:

$$L = 2^k$$

* **`k = 1` bit:** $L = 2^1 = 2$ levels. This is a **binary** (black and white) image. The intensity is either 0 or 1.
* **`k = 2` bits:** $L = 2^2 = 4$ levels (e.g., 0, 1, 2, 3).
* **`k = 8` bits:** $L = 2^8 = 256$ levels. This is the **standard for grayscale images**. The levels are 0, 1, 2, ..., 255.
* **`k = 24` bits (Color):** This is 8 bits for Red, 8 bits for Green, and 8 bits for Blue (8+8+8=24). This gives $2^{24} \approx 16.7$ million possible color combinations.

### Visual Example: The Effect of Poor Quantization

What happens when we reduce the number of intensity levels, `L` (by reducing the bit depth `k`)? 

We lose tonal detail. Smooth gradients (like the background of the skull X-ray) are forced into a few discrete steps. This creates visible, artificial edges called **false contouring** or **banding**.

## Image Storage

The total (uncompressed) storage size of an image is a simple product of its sampling and quantization (from *PDI-Aula01Parte_1-1, Slide 35*).

Let's say an image has `M` rows and `N` columns, and is quantized to `k` bits per pixel:

**Total bits = `M * N * k`**

* For a `1024 x 1024` **grayscale** (8-bit) image: `k = 8`.
    * Size = `1024 * 1024 * 8` = 8,388,608 bits.
* For a `1024 x 1024` **color** (24-bit) image: `k = 24` (8 bits * 3 channels).
    * Size = `1024 * 1024 * 24` = 25,165,824 bits.

This is why **image compression** (like JPEG or PNG) is so important! An uncompressed 12-megapixel (4000x3000) 24-bit color photo would be `4000 * 3000 * 24` = 288,000,000 bits, or over 34 megabytes!

## 🧠 Interactive Check / Reflection

A standard 1080p Full HD video frame has dimensions of `1920 x 1080` pixels. Assume it's a 24-bit color image.

1.  How many megabytes (MB) would **one single frame** take, uncompressed?
2.  A video plays at 30 frames per second. How many *gigabytes* (GB) would **one minute** of this uncompressed video take?

*(Remember: 8 bits = 1 byte, 1024 bytes = 1 KB, 1024 KB = 1 MB, 1024 MB = 1 GB)*

```python
# 1. Calculate storage for one frame
M = 1920
N = 1080
k = 24 # 24 bits (8 bits * 3 channels)

total_bits_frame = M * N * k
total_bytes_frame = total_bits_frame / 8
total_mb_frame = total_bytes_frame / (1024 * 1024)

print(f"One 1080p frame (uncompressed) is: {total_mb_frame:.2f} MB")

# 2. Calculate storage for one minute (30 fps * 60 seconds)
frames_per_second = 30
seconds = 60
total_frames = frames_per_second * seconds

total_mb_video = total_mb_frame * total_frames
total_gb_video = total_mb_video / 1024

print(f"One minute of 1080p/30fps video (uncompressed) is: {total_gb_video:.2f} GB")
```

*(That calculation should make it very clear why video streaming relies on heavy compression!)*

## 📜 Summary

* **Sampling** is the process of discretizing **space** (`x, y`) into a grid, which determines the **Spatial Resolution** (e.g., `1920x1080`).
* **Quantization** is the process of discretizing **intensity** (`f`) into a set of levels, which is determined by the **Bit Depth** (e.g., `8-bit` or 256 levels).
* Low spatial resolution causes "blockiness."
* Low intensity resolution causes "banding" or "false contours."
